```python
# Public-repository path setup.
# Run from anywhere inside the repository, or set CEFTAZIDIME_PROJECT_ROOT.
import os
from pathlib import Path

def _repo_root():
    env = os.environ.get("CEFTAZIDIME_PROJECT_ROOT")
    if env:
        return Path(env).expanduser().resolve()
    here = Path.cwd().resolve()
    for candidate in (here, *here.parents):
        if (candidate / "README.md").exists() and (candidate / "03_Notebooks").exists():
            return candidate
    return here

def _previous_project_root(project_root):
    env = os.environ.get("GENOME_MIC_AMR_PROJECT_ROOT")
    if env:
        return Path(env).expanduser().resolve()
    return (project_root / "external" / "Genome_MIC_AMR_Emergence").resolve()

PROJECT_ROOT = _repo_root()

#@title Cell F3.1 - Overview and paths
# Purpose:
# Create the two manuscript panels for Figure 3.
#
# Figure 3A:
# Collective chromosomal association:
# - previous genome-wide SNP K
# - whole-chromosome unitig K
#
# Figure 3B:
# Individual sequence-pattern screen:
# - 504,889 unique carrier patterns
# - BH-adjusted q-values
#
# Figure rule:
# Keep text inside the figure minimal.
# No explanatory legend is used when the caption can provide that information.

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
PROJECT_ROOT = _repo_root()

RESULTS_TABLE_DIR = (
    PROJECT_ROOT
    / "05_Results"
    / "Tables"
)

FIGURE_DIR = (
    PROJECT_ROOT
    / "06_Manuscript"
    / "Figures"
)

FIGURE_DIR.mkdir(parents=True, exist_ok=True)

NB18_SUMMARY = (
    RESULTS_TABLE_DIR
    / "18_collective_unitig_association_summary.csv"
)

FIG3A_PNG = FIGURE_DIR / "Figure3A_collective_chromosomal_association.png"
FIG3A_PDF = FIGURE_DIR / "Figure3A_collective_chromosomal_association.pdf"

FIG3B_PNG = FIGURE_DIR / "Figure3B_individual_sequence_pattern_screen.png"
FIG3B_PDF = FIGURE_DIR / "Figure3B_individual_sequence_pattern_screen.pdf"

assert PROJECT_ROOT.exists(), f"Project root not found: {PROJECT_ROOT}"
assert NB18_SUMMARY.exists(), f"Notebook 18 summary not found: {NB18_SUMMARY}"

print("Figure 3 manuscript notebook")
print("Figure output folder:", FIGURE_DIR)

print("\nTransition: Cell F3.2 will create Figure 3A.")
```

```python
#@title Cell F3.2 - Figure 3A collective chromosomal association
# Purpose:
# Compare the previously established SNP-relatedness association with the
# independently reconstructed whole-chromosome unitig association.
#
# Minimal manuscript presentation:
# - two bars
# - one y-axis
# - no legend
# - no explanatory text inside the panel

summary18 = pd.read_csv(NB18_SUMMARY)

assert len(summary18) == 1, "Expected one summary row in Notebook 18 output."

required_columns = [
    "previous_SNP_K_variance_fraction_reproduced",
    "whole_sequence_unitig_variance_fraction",
    "empirical_p_value",
]

for column in required_columns:
    assert column in summary18.columns, (
        f"Missing required Notebook 18 column: {column}\n"
        f"Available columns: {list(summary18.columns)}"
    )

snp_fraction = float(
    summary18.loc[0, "previous_SNP_K_variance_fraction_reproduced"]
)

unitig_fraction = float(
    summary18.loc[0, "whole_sequence_unitig_variance_fraction"]
)

unitig_p = float(
    summary18.loc[0, "empirical_p_value"]
)

assert np.isclose(snp_fraction, 0.611142, atol=1e-5)
assert np.isclose(unitig_fraction, 0.538815, atol=1e-5)
assert np.isclose(unitig_p, 0.000999, atol=1e-6)

values = np.array([snp_fraction, unitig_fraction], dtype=float)
labels = ["SNP K", "Unitig K"]

fig, ax = plt.subplots(figsize=(4.8, 4.6))

x = np.arange(len(values))

ax.bar(
    x,
    values,
    width=0.58,
    facecolor=["white", "black"],
    edgecolor="black",
    linewidth=1.0,
)

ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.set_ylabel("Variance fraction")
ax.set_ylim(0, 0.70)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

fig.tight_layout()

fig.savefig(FIG3A_PNG, dpi=300, bbox_inches="tight")
fig.savefig(FIG3A_PDF, bbox_inches="tight")

plt.show()
plt.close(fig)

print("Figure 3A created.")
print("SNP K variance fraction:", snp_fraction)
print("Unitig K variance fraction:", unitig_fraction)
print("Unitig permutation p:", unitig_p)
print(FIG3A_PNG)

print(
    "\nTransition: Cell F3.3 will locate the authoritative Notebook 17 "
    "pattern-association table and create Figure 3B."
)
```

```python
#@title Cell F3.2 - Figure 3A collective chromosomal association
# Purpose:
# Compare the previously established SNP-relatedness association with the
# independently reconstructed whole-chromosome unitig association.
#
# Manuscript presentation:
# - SNP K shown in white
# - Unitig K shown in gray
# - variance fractions shown above bars
# - no legend

summary18 = pd.read_csv(NB18_SUMMARY)

assert len(summary18) == 1, (
    "Expected one summary row in Notebook 18 output."
)

required_columns = [
    "previous_SNP_K_variance_fraction_reproduced",
    "whole_sequence_unitig_variance_fraction",
    "empirical_p_value",
]

for column in required_columns:
    assert column in summary18.columns, (
        f"Missing required Notebook 18 column: {column}\n"
        f"Available columns: {list(summary18.columns)}"
    )

snp_fraction = float(
    summary18.loc[
        0,
        "previous_SNP_K_variance_fraction_reproduced",
    ]
)

unitig_fraction = float(
    summary18.loc[
        0,
        "whole_sequence_unitig_variance_fraction",
    ]
)

unitig_p = float(
    summary18.loc[
        0,
        "empirical_p_value",
    ]
)

assert np.isclose(
    snp_fraction,
    0.611142,
    atol=1e-5,
)

assert np.isclose(
    unitig_fraction,
    0.538815,
    atol=1e-5,
)

assert np.isclose(
    unitig_p,
    0.000999,
    atol=1e-6,
)

values = np.array(
    [
        snp_fraction,
        unitig_fraction,
    ],
    dtype=float,
)

labels = [
    "SNP K",
    "Unitig K",
]

fig, ax = plt.subplots(
    figsize=(4.8, 4.6)
)

x = np.arange(
    len(values)
)

bars = ax.bar(
    x,
    values,
    width=0.58,
    facecolor=[
        "white",
        "0.65",
    ],
    edgecolor="black",
    linewidth=1.0,
)

# Show variance fractions above the bars.
for bar, value in zip(bars, values):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        value + 0.015,
        f"{value:.3f}",
        ha="center",
        va="bottom",
        fontsize=10,
    )

ax.set_xticks(x)
ax.set_xticklabels(labels)

ax.set_ylabel(
    "Variance fraction"
)

ax.set_ylim(
    0,
    0.70,
)

ax.spines[
    "top"
].set_visible(False)

ax.spines[
    "right"
].set_visible(False)

fig.tight_layout()

fig.savefig(
    FIG3A_PNG,
    dpi=300,
    bbox_inches="tight",
)

fig.savefig(
    FIG3A_PDF,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)

print("Figure 3A created.")
print(
    "SNP K variance fraction:",
    f"{snp_fraction:.3f}",
)
print(
    "Unitig K variance fraction:",
    f"{unitig_fraction:.3f}",
)
print(
    "Unitig permutation p:",
    unitig_p,
)
print(
    "Saved PNG:",
    FIG3A_PNG,
)
print(
    "Saved PDF:",
    FIG3A_PDF,
)
```

```python
#@title Cell F3.3 - Figure 3B individual sequence-pattern association
# Purpose:
# Show all 504,889 individual sequence-pattern association results.
#
# Three patterns survive Benjamini-Hochberg correction at q <= 0.05.
#
# Figure rule:
# Minimal text, grayscale-compatible, no legend.

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

EXPECTED_PATTERNS = 504_889

NB17_DIR = (
    PROJECT_ROOT
    / "04_Intermediate"
    / "17_Full_Chromosomal_Unitig_Pattern_Association"
)

ALL_PATTERN_RESULTS = (
    NB17_DIR
    / "17_all_unique_pattern_association_results.csv.gz"
)

FIG3B_PNG = (
    FIGURE_DIR
    / "Figure3B_individual_sequence_pattern_screen.png"
)

FIG3B_PDF = (
    FIGURE_DIR
    / "Figure3B_individual_sequence_pattern_screen.pdf"
)

assert ALL_PATTERN_RESULTS.exists()

pattern_results = pd.read_csv(
    ALL_PATTERN_RESULTS
)

assert len(
    pattern_results
) == EXPECTED_PATTERNS

Q_COLUMN = "Benjamini_Hochberg_q_value"

assert Q_COLUMN in pattern_results.columns

q_values = pd.to_numeric(
    pattern_results[
        Q_COLUMN
    ],
    errors="coerce",
).to_numpy(
    dtype=float
)

assert np.isfinite(
    q_values
).all()

significant_mask = (
    q_values <= 0.05
)

significant_count = int(
    significant_mask.sum()
)

assert significant_count == 3, (
    f"Expected 3 BH-significant patterns, found {significant_count}."
)

# ------------------------------------------------------------
# Rank all patterns by q-value
# ------------------------------------------------------------

sort_order = np.argsort(
    q_values
)

sorted_q = q_values[
    sort_order
]

safe_q = np.maximum(
    sorted_q,
    np.finfo(float).tiny,
)

minus_log10_q = (
    -np.log10(
        safe_q
    )
)

rank = np.arange(
    1,
    EXPECTED_PATTERNS + 1
)

threshold = (
    -np.log10(
        0.05
    )
)

# ------------------------------------------------------------
# Plot ranked BH q-values as points
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(6.8, 4.6)
)

ax.scatter(
    rank,
    minus_log10_q,
    s=2,
    facecolors="none",
    edgecolors="black",
    linewidths=0.25,
    rasterized=True,
)

ax.axhline(
    threshold,
    linestyle="--",
    color="black",
    linewidth=0.9,
)

# Three BH-significant patterns.
ax.scatter(
    rank[:3],
    minus_log10_q[:3],
    s=34,
    facecolors="black",
    edgecolors="black",
)

ax.set_xscale(
    "log"
)

ax.set_xlabel(
    "Pattern rank"
)

ax.set_ylabel(
    "−log10(BH q)"
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

fig.tight_layout()

fig.savefig(
    FIG3B_PNG,
    dpi=300,
    bbox_inches="tight",
)

fig.savefig(
    FIG3B_PDF,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)

# ------------------------------------------------------------
# QC
# ------------------------------------------------------------

significant_patterns = (
    pattern_results.loc[
        significant_mask,
        [
            "pattern_id",
            "present_count",
            "high_MIC_count",
            "remaining_count",
            "estimated_log2_MIC_difference",
            "p_value_two_sided",
            "Benjamini_Hochberg_q_value",
            "estimated_direction",
        ],
    ]
    .sort_values(
        "Benjamini_Hochberg_q_value"
    )
    .reset_index(
        drop=True
    )
)

print(
    "Figure 3B created."
)

print(
    "Patterns tested:",
    f"{EXPECTED_PATTERNS:,}"
)

print(
    "Patterns with BH q <= 0.05:",
    significant_count
)

display(
    significant_patterns
)

print(
    FIG3B_PNG
)
```

```python
#@title Cell F3.3a - Find the authoritative full Notebook 17 association output
# Purpose:
# Inspect Notebook 17 itself and identify exactly where the full
# 504,889-pattern association results were saved.
#
# No file path is guessed.

from pathlib import Path
import json
import re

PROJECT_ROOT = _repo_root()

notebook_candidates = list(
    PROJECT_ROOT.rglob(
        "17_Full_Chromosomal_Unitig_Pattern_Association_with_Continuous_Ceftazidime_MIC.ipynb"
    )
)

assert len(notebook_candidates) == 1, (
    "Expected exactly one Notebook 17 file.\n"
    f"Found: {notebook_candidates}"
)

NB17_FILE = notebook_candidates[0]

print("Notebook 17:")
print(NB17_FILE)

with open(
    NB17_FILE,
    "r",
    encoding="utf-8",
) as handle:
    notebook = json.load(handle)

keywords = [
    "to_csv",
    "to_parquet",
    "savez",
    "savez_compressed",
    ".npz",
    ".csv",
    ".csv.gz",
    ".parquet",
    "association",
    "q_value",
    "Benjamini",
]

print("\nRelevant Notebook 17 code lines:\n")

for cell_number, cell in enumerate(
    notebook.get("cells", []),
    start=1,
):
    if cell.get("cell_type") != "code":
        continue

    source = "".join(
        cell.get(
            "source",
            []
        )
    )

    matching_lines = []

    for line in source.splitlines():
        if any(
            keyword.lower() in line.lower()
            for keyword in keywords
        ):
            matching_lines.append(
                line
            )

    if matching_lines:
        print(
            f"--- Notebook cell {cell_number} ---"
        )

        for line in matching_lines:
            print(
                line
            )

        print()

print(
    "Diagnostic complete. "
    "Please send me this output before we change Figure 3B."
)
```

```python
#@title Cell F3.4 - Combine Figure 3A and Figure 3B
# Purpose:
# Combine the final Figure 3A and Figure 3B PNG files into one manuscript Figure 3.
#
# Layout:
# - panel A on the left
# - panel B on the right
# - bold panel labels A and B
# - no additional legend or interpretation added
#
# The existing Figure 3A and Figure 3B plotting cells are not changed.

from PIL import Image, ImageDraw, ImageFont

# ------------------------------------------------------------
# Verify panel files
# ------------------------------------------------------------

for path in [
    FIG3A_PNG,
    FIG3B_PNG,
]:
    assert path.exists(), (
        f"Required panel image not found: {path}"
    )

# ------------------------------------------------------------
# Load panels
# ------------------------------------------------------------

panel_a = Image.open(
    FIG3A_PNG
).convert(
    "RGB"
)

panel_b = Image.open(
    FIG3B_PNG
).convert(
    "RGB"
)

# Use the smaller panel height so neither image is upscaled.
target_height = min(
    panel_a.height,
    panel_b.height,
)

def resize_to_height(
    image,
    height,
):
    width = int(
        round(
            image.width
            * height
            / image.height
        )
    )

    return image.resize(
        (
            width,
            height,
        ),
        Image.Resampling.LANCZOS,
    )

panel_a = resize_to_height(
    panel_a,
    target_height,
)

panel_b = resize_to_height(
    panel_b,
    target_height,
)

# ------------------------------------------------------------
# Create combined canvas
# ------------------------------------------------------------

outer_margin = max(
    30,
    int(
        round(
            target_height
            * 0.035
        )
    ),
)

top_margin = max(
    55,
    int(
        round(
            target_height
            * 0.065
        )
    ),
)

panel_gap = max(
    35,
    int(
        round(
            target_height
            * 0.035
        )
    ),
)

canvas_width = (
    outer_margin
    + panel_a.width
    + panel_gap
    + panel_b.width
    + outer_margin
)

canvas_height = (
    top_margin
    + target_height
    + outer_margin
)

combined = Image.new(
    "RGB",
    (
        canvas_width,
        canvas_height,
    ),
    "white",
)

x_a = outer_margin
x_b = (
    outer_margin
    + panel_a.width
    + panel_gap
)
y_panel = top_margin

combined.paste(
    panel_a,
    (
        x_a,
        y_panel,
    ),
)

combined.paste(
    panel_b,
    (
        x_b,
        y_panel,
    ),
)

# ------------------------------------------------------------
# Add panel labels
# ------------------------------------------------------------

font_size = max(
    28,
    int(
        round(
            target_height
            * 0.045
        )
    ),
)

try:
    font = ImageFont.truetype(
        "DejaVuSans-Bold.ttf",
        font_size,
    )
except OSError:
    font = ImageFont.load_default()

draw = ImageDraw.Draw(
    combined
)

label_y = max(
    5,
    int(
        round(
            top_margin
            * 0.15
        )
    ),
)

draw.text(
    (
        x_a,
        label_y,
    ),
    "A",
    fill="black",
    font=font,
)

draw.text(
    (
        x_b,
        label_y,
    ),
    "B",
    fill="black",
    font=font,
)

# ------------------------------------------------------------
# Save combined Figure 3
# ------------------------------------------------------------

FIG3_COMBINED_PNG = (
    FIGURE_DIR
    / "Figure3_combined.png"
)

FIG3_COMBINED_PDF = (
    FIGURE_DIR
    / "Figure3_combined.pdf"
)

combined.save(
    FIG3_COMBINED_PNG,
    dpi=(
        300,
        300,
    ),
)

combined.save(
    FIG3_COMBINED_PDF,
    resolution=300.0,
)

display(
    combined
)

print(
    "Combined Figure 3 created."
)

print(
    "Saved PNG:",
    FIG3_COMBINED_PNG,
)

print(
    "Saved PDF:",
    FIG3_COMBINED_PDF,
)
```
